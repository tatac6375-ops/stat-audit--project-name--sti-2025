## Analisis Inferensia: Pengujian Hipotesis (Z-Test)

```python
import sys
sys.path.append("../")
import numpy as np
from src.hypothesis import z_test_one_sample

# Data hulu yang diserahterimakan dari Member B dan Member C
n = 4042
x_bar = 0.62098  # Nilai point estimate proporsi sukses merge PR

# Standard deviasi untuk distribusi proporsi (Bernoulli)
sigma = np.sqrt(x_bar * (1 - x_bar))

# Nilai komparasi hipotesis awal (Klaim historis 60%)
mu0 = 0.60000 

# Eksekusi fungsi pengujian hipotesis dari src/hypothesis.py (Sesuai Tsun 2020 p. 306)
hasil = z_test_one_sample(
    x_bar=x_bar, 
    mu0=mu0, 
    sigma=sigma, 
    n=n, 
    alternative='two-sided', 
    alpha=0.05
)

print("=== LIVE REPORT CHECKPOINT ===")
print("Z-Statistic :", round(hasil["z_stat"], 4))
print("P-Value     :", round(hasil["p_value"], 4))
print("Keputusan   :", hasil["decision"])
print("Narasi      :", hasil["interpretation"])
```

### Prosedur Uji Hipotesis (6-Step Procedure) - Member D

**Step 1: State Hypotheses**
* $H_0: \mu = 0.60$ (Proporsi dari total pull request yang berhasil di-merge pada repositori Pandas adalah sama dengan angka historis 60%)
* $H_a: \mu \neq 0.60$ (Proporsi dari total pull request yang berhasil di-merge pada repositori Pandas secara signifikan tidak sama dengan angka historis 60%)

**Step 2: Assumptions Check**
* Data pull request yang digunakan bersifat independen karena ditarik langsung dari log aktivitas sistem Git repositori `pandas-dev/pandas`.
* Ukuran sampel data yang kita miliki sangat besar yaitu $n = 4042$, di mana nilai $n \cdot p \ge 10$ dan $n \cdot (1-p) \ge 10$ terpenuhi secara mutlak. Berdasarkan Teorema Limit Pusat (Central Limit Theorem), bentuk distribusi sampling dari proporsi ini akan berkonvergensi mendekati distribusi normal standar. Oleh karena itu, penggunaan uji satu sampel Z-Test ini dinilai valid.

**Step 3: Test Statistic Calculation**
* Formula konseptual komputasi merujuk pada Tsun (2020) halaman 306:
  $$Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$$
* Melalui eksekusi fungsi `z_test_one_sample` pada modul `src/hypothesis.py`, diperoleh nilai Z-statistic sebesar **2.7494**.

**Step 4: P-value and Critical Region**
* Pengujian ini menggunakan tingkat signifikansi $\alpha = 0.05$ dengan tipe dua arah (two-sided).
* Wilayah kritis untuk menolak $H_0$ ditentukan pada batas nilai $Z < -1.96$ atau $Z > 1.96$.

**Step 5: Statistical Decision**
* Berdasarkan perhitungan komputasi dengan `scipy.stats`, diperoleh nilai p-value sebesar **0.0060**.
* Karena nilai p-value ($0.0060$) < $\alpha$ ($0.05$) dan nilai $Z$-stat ($2.7494$) berada di dalam wilayah kritis ($Z > 1.96$), maka keputusan statistik yang diambil secara tegas adalah **Reject $H_0$** (Menolak $H_0$).

**Step 6: Conclusion & Interpretation**
* Dengan tingkat signifikansi 5%, terdapat bukti statistik yang sangat kuat untuk menolak $H_0$. Hal ini menyimpulkan bahwa proporsi pull request yang berhasil masuk ke repositori Pandas berbeda secara nyata dari klaim standar 60%. Performa riil yang berada di angka 62.09% menunjukkan tingkat penerimaan kode yang lebih tinggi. Bagi maintainer, ini adalah sinyal positif bahwa alur kerja *code review* mereka berjalan produktif dan kontribusi komunitas tersaring dengan baik tanpa penumpukan *backlog*.
